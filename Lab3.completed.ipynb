#### 7. Scale your randomly generated numbers to be between 0 and 10.

```python
rand_arr = rand_arr * 10
```

#### 8. Load the data contained in the "Random_Data.csv" file from Canvas directly as an ndarray.

```python
loaded_data = np.loadtxt("Random_Data.csv", delimiter=",")
```

#### 9. Find the difference between your randomly generated data and the loaded data. Save it as an ndarray called diff_array.

```python
diff_array = rand_arr - loaded_data
```

#### 10. Identify the lowest and highest values in diff_array.

```python
min_val = diff_array.min()
max_val = diff_array.max()
print("Lowest:", min_val, "Highest:", max_val)
```

#### 11. Identify the location of the lowest and highest values in diff_array.

```python
min_loc = np.unravel_index(diff_array.argmin(), diff_array.shape)
max_loc = np.unravel_index(diff_array.argmax(), diff_array.shape)
print("Lowest Location (row, col):", min_loc)
print("Highest Location (row, col):", max_loc)
```

#### 12. Select the row containing the lowest value in diff_array. Save it as an ndarray called low_array.

```python
low_row_idx = min_loc[0]
low_array = diff_array[low_row_idx, :]
```

#### 13. Select the column containing the highest value in diff_array. Save it as an ndarray called high_array.

```python
high_col_idx = max_loc[1]
high_array = diff_array[:, high_col_idx]
```

#### 14. Turn both high_array and low_array into 10 x 10 arrays.

```python
low_array = low_array.reshape(10, 10)
high_array = high_array.reshape(10, 10)
```

#### 15. Calculate the dot product of high_array and low_array.

```python
dot_product = np.dot(high_array, low_array)
```

#### 16. Merge the two arrays together to form a singular 10 x 20 array. Save this as an ndarray called fusion_array.

```python
fusion_array = np.hstack((high_array, low_array))
```

#### 17. Add a column of all 0s in the middle of fusion_array.

```python
zero_col = np.zeros((10, 1))
fusion_array = np.insert(fusion_array, 10, zero_col.squeeze(), axis=1)
```

#### 18. Replace all numbers in fusion_array less than 0 with 0.

```python
fusion_array[fusion_array < 0] = 0
```

#### 19. Print the number of values in fusion_array greater than 0.

```python
print(np.sum(fusion_array > 0))
```

#### 20. Save fusion_array as a text file called 'Random_Fusion.tsv'.

```python
np.savetxt('Random_Fusion.tsv', fusion_array, delimiter='\t')
```
